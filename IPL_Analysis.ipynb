```python
# Death Overs Performance

death_overs = df[df["over"]>=16]
```

```python
#  Runs scored in death overs by Batters
death_runs = death_overs.groupby("batter")["batsman_runs"].sum()
```

```python
# Balls faced by batters in death overs
death_balls = death_overs.groupby("batter")["ball"].count()
```

```python
# Batsmen strike rate in death overs 

death_sr = (( death_runs / death_balls ) * 100 ).round(2)
```
